```python
#Soal no 1

n = int(input("Masukkan bilangan bulat (1-5): "))

if n == 1:
    faktorial = 1
elif n == 2:
    faktorial = 2 * 1
elif n == 3:
    faktorial = 3 * 2 * 1
elif n == 4:
    faktorial = 4 * 3 * 2 * 1
elif n == 5:
    faktorial = 5 * 4 * 3 * 2 * 1
else:
    faktorial = 0
    print("Input tidak valid! Masukkan angka 1 sampai 5.")

if n >= 1 and n <= 5:
    print("Faktorial dari", n, "adalah", faktorial)

# Soal no 2

faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

skor = faktor_daya * efisiensi

print(f"Skor kualitas daya: {skor:.2f}")

if skor >= 85 and faktor_daya >= 0.90:
    klasifikasi = "PRIMA"
elif skor >= 70 and faktor_daya >= 0.80:
    klasifikasi = "BAIK"
elif skor >= 50 or faktor_daya >= 0.75:
    klasifikasi = "CUKUP"
else:
    klasifikasi = "BURUK"

print("Klasifikasi:", klasifikasi)


# Soal no 3

iradiasi = float(input("Masukkan nilai iradiasi matahari (W/m²): "))
soc = float(input("Masukkan State of Charge (SoC) baterai (%): "))


daya_panel = float(input("Masukkan estimasi daya panel (W): "))

print(f"Estimasi daya panel: {daya_panel:.1f} W")

# Menentukan mode operasi PLTS
if iradiasi > 600 and soc < 80:
    mode = "CHARGING — prioritaskan pengisian baterai"

elif iradiasi > 600 and soc >= 80:
    mode = "GRID-EXPORT — ekspor kelebihan daya ke jaringan"

elif iradiasi > 200 and soc >= 20:
    mode = "NORMAL — operasi mandiri"

elif iradiasi <= 200 and soc >= 50:
    mode = "BATTERY-ONLY — andalkan baterai"

else:
    mode = "EMERGENCY — kurangi beban segera"

print(f"Iradiasi {iradiasi} W/m², SoC {soc}% → Mode: {mode}")
```
